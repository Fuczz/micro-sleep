# Train YOLO — tách yaml mắt / ngáp, 50–100 epoch

Không dùng chung `dataset.yaml` cho cả hai bài. Sau train, mở `confusion_matrix.png` trong `runs/`.

```python
from ultralytics import YOLO

EPOCHS = 80
BATCH = 16
IMGSZ = 320

eye = YOLO("yolov8n.pt")
eye.train(data="dataset_eye.yaml", epochs=EPOCHS, batch=BATCH, imgsz=IMGSZ,
          project="runs/detecteye", name="train", exist_ok=True, plots=True)
eye.val(data="dataset_eye.yaml", split="test", plots=True)
```
